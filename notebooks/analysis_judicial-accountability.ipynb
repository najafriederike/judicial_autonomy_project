### 2 Subset Datasets

```python
# Subset Q1
# Filter for sub-indices, indicators and index
# country-year information: 'country', 'country_code', 'year', 'country_year'
# sub-indices judges rights: 'q1c2_tenure', 'q1c2_term', 'q1c2_immunity', 'q1c2_transfer', 'q1c2_finance'
# sub-indices judges obligations: 'q1c3_proced', 'q1c3_actor', 'q1c3_conflict'
# indicators: 'q1_select', 'q1_admin', 'q1_right', 'q1_oblig'
# index:'q1JA_wm', 'q1JA_wp'

q1_subset = q1[['country', 'country_code', 'year', 'country_year',
                'q1c2_tenure', 'q1c2_term', 'q1c2_immunity', 'q1c2_transfer', 
                'q1c2_finance', 'q1c3_proced', 'q1c3_actor', 'q1c3_conflict',
                'q1_select', 'q1_admin', 'q1_right', 'q1_oblig',
                'q1JA_wm', 'q1JA_wp']]
q1_subset.shape
```

```python
# Subset Q2
# Filter for sub-indices, indicators and index 
# country-year information: 'country', 'country_code', 'year', 'country_year'
# sub-indices judges rights: 'q2c2_term', 'q2c2_immunity', 'q2c2_removal', 'q2c2_finance'
# sub-indices judges obligations: 'q2c3_proced', 'q2c3_actor', 'q2c3_conflict'
# indicators: 'q2_select', 'q2_admin', 'q2_right', 'q2_oblig'
# index: 'q2JA_wm', 'q2JA_wp'

q2_subset = q2[['country', 'country_code', 'year', 'country_year',
                'q2c2_term', 'q2c2_immunity', 'q2c2_removal', 'q2c2_finance', 
                'q2c3_proced', 'q2c3_actor', 'q2c3_conflict',
                'q2_select', 'q2_admin', 'q2_right', 'q2_oblig',
                'q2JA_wm', 'q2JA_wp']]
q2_subset.shape
```

```python
# Merge datasets

q1_q2_merge = pd.merge(q1_subset, q2_subset, on=['country', 'country_code', 'year', 'country_year'])
q1_q2_merge.head()
```

```python
# Check for missing values

q1_q2_merge.isna().sum()
```

#### Measures of central tendency

#### Histograms

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

##### _Ordinary Courts_

```python
# Plot distribution of judges’ obligation scores for ordinary courts

# Creating customized histogram with a density plot
sns.histplot(
    q1_q2_merge['q1_oblig'], 
    bins=30, 
    color='#7FBCE6', 
    edgecolor='#003A70', 
    linewidth=0.8, 
    kde=True)

# Labels and title
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram: Judges Obligations (Ordinary Courts)')

# Display plot
plt.show()
```

```python
# Plot distribution of judges’ obligation scores for ordinary courts (incl. mean, median and mode)

# Calculate summary statistics
mean_val = q1_q2_merge['q1_oblig'].mean()
median_val = q1_q2_merge['q1_oblig'].median()
mode_val = q1_q2_merge['q1_oblig'].mode()[0]

# Plot histogram with KDE
sns.histplot(
    q1_q2_merge['q1_oblig'],
    bins=30,
    color='#7FBCE6',
    edgecolor='#003A70',
    linewidth=0.8,
    kde=True)

# Add vertical lines for mean, median, and mode
plt.axvline(mean_val, color='#8B1E1E', linestyle='-', linewidth=1, label='Mean')
plt.axvline(median_val, color='#B03A2E', linestyle='--', linewidth=1, label='Median')
plt.axvline(mode_val, color='#D98880', linestyle=':', linewidth=1, label='Mode')

# Labels and title
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram: Judges’ Obligations (Ordinary Courts)')

# Legend
plt.legend()

# Display the plot
plt.show()
```

##### _Ordinary vs. Apex Courts_

```python
# Plot distribution of judges’ obligation scores for ordinary and apex courts

# Creating a stacked histogram
plt.hist(
    [q1_q2_merge['q1_oblig'], q1_q2_merge['q2_oblig']], 
    bins=30, 
    stacked=True, 
    color=['#7FBCE6', '#003A70'], 
    edgecolor='#003A70',
    alpha=0.7)

# Adding labels and title
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.title('Histogram: Judges Obligations')

# Adding legend
plt.legend(['Ordinary Courts', 'Apex Courts'])

# Display the plot
plt.show()
```
